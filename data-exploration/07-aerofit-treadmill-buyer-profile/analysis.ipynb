```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/aerofit_treadmill_data.csv")
print(df.shape)
print(df.dtypes)
print(df.head())
print(df.isnull().sum())
print(df.describe())
print(df.describe(include="object"))
```

```python
# Step 2 now — product mix (exact counts)


print(df["Product"].value_counts())
print(df["Product"].value_counts(normalize=True).round(3))
```

```python
# Step 3 — First two contingency tables (25 min)


# counts
gender_ct = pd.crosstab(df["Product"], df["Gender"], margins=True)
print(gender_ct)

# row %  = P(Gender | Product)
print(pd.crosstab(df["Product"], df["Gender"], normalize="index").round(3))

# column % = P(Product | Gender)
print(pd.crosstab(df["Product"], df["Gender"], normalize="columns").round(3))
```

```python
# Block A — remaining contingency tables
```

```python
def show_crosstab(col):
    print("=" * 60)
    print(col)
    print("COUNTS")
    print(pd.crosstab(df["Product"], df[col], margins=True))
    print("\nP(col | Product)  normalize='index'")
    print(pd.crosstab(df["Product"], df[col], normalize="index").round(3))
    print("\nP(Product | col)  normalize='columns'")
    print(pd.crosstab(df["Product"], df[col], normalize="columns").round(3))

show_crosstab("MaritalStatus")
show_crosstab("Fitness")
show_crosstab("Usage")
```

Block B — numeric EDA

```python
num_cols = ["Age", "Education", "Usage", "Fitness", "Income", "Miles"]

print(df.groupby("Product")[num_cols].median())
print(df.groupby("Product")[num_cols].mean().round(1))

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
sns.boxplot(data=df, x="Product", y="Income", ax=axes[0, 0])
sns.boxplot(data=df, x="Product", y="Miles", ax=axes[0, 1])
sns.boxplot(data=df, x="Product", y="Age", ax=axes[1, 0])
sns.boxplot(data=df, x="Product", y="Fitness", ax=axes[1, 1])
plt.tight_layout()
plt.show()

sns.heatmap(df[num_cols].corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation of numeric features")
plt.show()
```
